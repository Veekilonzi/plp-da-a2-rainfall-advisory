
# Rainfall Analysis and Planting Advisory

## Project Overview

This project analyses monthly rainfall data from five Kenyan counties using Python and NumPy. It identifies annual rainfall totals, monthly averages, wettest months, dry months, and recommended two-month planting windows.

The analysis also cleans invalid Machakos rainfall readings and documents the imputation process.

## Counties Covered

- Nairobi
- Nakuru
- Kisumu
- Mombasa
- Machakos


```python
import numpy as np
import csv
```

```python

# List of counties
counties = ["Nairobi", "Nakuru", "Kisumu", "Mombasa", "Machakos"]

# Rainfall data for the first four counties
clean_rows = [
    [64, 56, 92, 219, 176, 35, 18, 24, 31, 61, 165, 104],  # Nairobi
    [35, 42, 70, 145, 120, 78, 96, 105, 72, 55, 60, 38],   # Nakuru
    [58, 90, 150, 240, 180, 95, 70, 85, 90, 88, 130, 95],  # Kisumu
    [30, 15, 55, 150, 240, 110, 90, 65, 60, 90, 100, 70]   # Mombasa
]

# Raw Machakos rainfall data
machakos_raw = [
    "48", "35", "88", "N/A", "140", " 20 ",
    "9", "12", "", "45", "182", "96"
]

# Month names
months = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

print("Counties:", counties)
print("Number of clean county rows:", len(clean_rows))
print("Number of Machakos readings:", len(machakos_raw))
```

```python
import numpy as np

def clean_machakos(data):
    valid_values = []
    invalid_months = []

    for index, value in enumerate(data):
        try:
            valid_values.append(int(value.strip()))
        except ValueError:
            invalid_months.append(months[index])

    # Calculate the rounded average of valid readings
    average = round(np.mean(valid_values))

    # Replace invalid readings with the average
    cleaned_values = []

    for value in data:
        try:
            cleaned_values.append(int(value.strip()))
        except ValueError:
            cleaned_values.append(average)

    print("Imputed months:", invalid_months)
    print("Imputed value:", average, "mm")

    return cleaned_values


machakos_clean = clean_machakos(machakos_raw)

print("Cleaned Machakos rainfall data:")
print(machakos_clean)
```

```python

# Add cleaned Machakos data to the other four counties
all_rows = clean_rows + [machakos_clean]

# Convert the combined data into a NumPy array
rain = np.array(all_rows)

# Display the array
print("Combined rainfall array:")
print(rain)

# Display shape and data type
print("\nArray shape:", rain.shape)
print("Array data type:", rain.dtype)
```

```python

# Calculate annual rainfall totals for each county
annual_totals = rain.sum(axis=1)

print("Annual rainfall totals:")
for county, total in zip(counties, annual_totals):
    print(f"{county}: {total} mm")
```

```python

# Calculate average rainfall for each month across all counties
monthly_averages = rain.mean(axis=0)

print("Average monthly rainfall across all counties:")

for month, average in zip(months, monthly_averages):
    print(f"{month}: {average:.2f} mm")
```

```python

# Find the index of the wettest month for each county
wettest_indices = np.argmax(rain, axis=1)

# Convert indices into month names
wettest_months = np.array(months)[wettest_indices]

print("Wettest month for each county:")

for county, month in zip(counties, wettest_months):
    print(f"{county}: {month}")
```

```python

# Create a boolean mask for rainfall below 50 mm
dry_mask = rain < 50

# Count dry months for each county
dry_month_counts = dry_mask.sum(axis=1)

print("Number of dry months per county:")

for county, count in zip(counties, dry_month_counts):
    print(f"{county}: {count} dry months")
```

```python

# Calculate combined rainfall for consecutive months
two_month_rainfall = rain[:, :-1] + rain[:, 1:]

# Find the index of the highest two-month rainfall
planting_indices = np.argmax(two_month_rainfall, axis=1)

# Get the corresponding month pairs
planting_windows = [
    f"{months[i]}-{months[i + 1]}"
    for i in planting_indices
]

print("Recommended two-month planting windows:")

for county, window in zip(counties, planting_windows):
    print(f"{county}: {window}")
```

```python

print("FINAL RAINFALL SUMMARY")
print("=" * 55)

for county, total, wettest, dry, window in zip(
    counties,
    annual_totals,
    wettest_months,
    dry_month_counts,
    planting_windows
):
    print(f"\nCounty: {county}")
    print(f"Annual rainfall: {total} mm")
    print(f"Wettest month: {wettest}")
    print(f"Dry months: {dry}")
    print(f"Planting window: {window}")
```

```python
import csv

# Create the rainfall summary CSV file
with open("rainfall_summary.csv", "w", newline="") as file:
    writer = csv.writer(file)

    # Write the column headings
    writer.writerow([
        "county",
        "annual_total_mm",
        "wettest_month",
        "dry_months",
        "planting_window"
    ])

    # Write the results for each county
    for row in zip(
        counties,
        annual_totals,
        wettest_months,
        dry_month_counts,
        planting_windows
    ):
        writer.writerow(row)

print("rainfall_summary.csv created successfully!")
```


## Planting Advisory

Based on the rainfall analysis, Nairobi, Nakuru, Kisumu, and Mombasa have their highest consecutive two-month rainfall during April and May, making this period a recommended planting window. Machakos has its highest combined rainfall during November and December, suggesting this period as a suitable planting window. Farmers should prepare their land before the recommended periods to take advantage of the expected rainfall patterns. The NGO should supplement these recommendations with local weather forecasts before sending planting advice. These recommendations are based on historical monthly readings and do not guarantee future rainfall conditions.



## Data Quality Notes

The original Machakos rainfall dataset contained two invalid entries: "N/A" for April and an empty value for September. These entries were identified using a try/except ValueError statement and replaced with the rounded mean of the valid Machakos readings, which was 68 mm. The imputation process allowed all five counties to be included in the NumPy analysis. However, the replacement values are estimates and may not represent the actual rainfall recorded during those months. The NGO should verify the original station records to improve the reliability of future planting recommendations.
